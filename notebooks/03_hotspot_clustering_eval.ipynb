```python
import pandas as pd
import numpy as np
from sklearn.metrics import silhouette_score

hotspots_df = pd.read_csv("../data/processed/hotspot_clusters_summary.csv")
clustered_pings = pd.read_csv("../data/processed/clustered_mobility_pings.csv")

noise_ratio = (clustered_pings["cluster_id"] == -1).mean() * 100
valid_clusters = clustered_pings[clustered_pings["cluster_id"] != -1]

coords_rad = np.radians(valid_clusters[["latitude", "longitude"]].values)
sil_score = silhouette_score(coords_rad, valid_clusters["cluster_id"], metric="haversine")

print("=== Nagpur District DBSCAN Hotspot Evaluation ===")
print(f"Total Discovered Hotspots: {len(hotspots_df)}")
print(f"Clustered Pings: {len(valid_clusters)} | Isolated Noise Filtered Out: {noise_ratio:.2f}%")
print(f"Haversine Silhouette Score: {sil_score:.4f}")
```

```python
import pandas as pd

enriched_df = pd.read_csv("../data/processed/enriched_mobility_pings.csv")
store_kpis = pd.read_csv("../data/processed/store_performance_kpis.csv")

print("=== Foot Traffic by Commercial Zone ===")
print(enriched_df["zone"].value_counts())

print("\n=== Catchment Ring Distribution ===")
print(enriched_df["catchment_ring"].value_counts(normalize=True).mul(100).round(2).astype(str) + "%")
```

```python
time_slot_summary = (
    enriched_df.groupby(["zone", "time_slot"])
    .agg(
        total_pings=("ping_id", "count"),
        avg_dwell_mins=("dwell_time_mins", "mean"),
        avg_spend_inr=("estimated_spend_inr", "mean"),
    )
    .round(2)
    .reset_index()
)
time_slot_summary.head(12)
```

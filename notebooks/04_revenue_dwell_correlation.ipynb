# 04 - Revenue & Dwell Time Correlation Analysis (Nagpur District)
Analyzes correlation between customer dwell duration (`dwell_time_mins`) and `estimated_spend_inr`, plus `spend_per_minute` efficiency across Nagpur retail corridors and categories.

```python
from pathlib import Path
import numpy as np
import pandas as pd
from scipy import stats

candidate_paths = [
    Path('data/processed/enriched_mobility_pings.csv'),
    Path('data/processed/cleaned_mobility_pings.csv'),
    Path('data/raw/mobility_pings.csv')
]
data_path = next((p for p in candidate_paths if p.exists()), None)

if data_path:
    df = pd.read_csv(data_path)
    print(f'Loaded {len(df):,} rows from {data_path}')
    print('Columns:', list(df.columns))
    if {'dwell_time_mins', 'estimated_spend_inr'}.issubset(df.columns):
        valid = df.dropna(subset=['dwell_time_mins', 'estimated_spend_inr']).copy()
        r_pear, p_pear = stats.pearsonr(valid['dwell_time_mins'], valid['estimated_spend_inr'])
        r_spear, p_spear = stats.spearmanr(valid['dwell_time_mins'], valid['estimated_spend_inr'])
        print(f'Pearson r : {r_pear:.4f} (p-value: {p_pear:.4e})')
        print(f'Spearman r: {r_spear:.4f} (p-value: {p_spear:.4e})')
        valid['spend_per_minute'] = valid['estimated_spend_inr'] / valid['dwell_time_mins'].clip(lower=1)
        for col in ['corridor', 'zone', 'market_zone', 'category', 'store_category']:
            if col in valid.columns:
                summary = valid.groupby(col).agg(
                    visits=('estimated_spend_inr', 'count'),
                    avg_dwell_mins=('dwell_time_mins', 'mean'),
                    avg_spend_inr=('estimated_spend_inr', 'mean'),
                    mean_spend_per_min=('spend_per_minute', 'mean'),
                    median_spend_per_min=('spend_per_minute', 'median')
                ).sort_values('mean_spend_per_min', ascending=False)
                print(f'\n--- Summary by {col} ---')
                display(summary)
                break
else:
    print('No mobility CSV found in expected data/ paths. Check data/ folder contents.')
```
